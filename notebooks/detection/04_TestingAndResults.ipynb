```python
from ultralytics import YOLO

model = YOLO("/content/best.pt")
print("Model loaded successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
results = model.predict(
    source="/content/img1.jpg",
    conf=0.25,
    save=True
)
```

```python
from IPython.display import Image, display

display(Image(filename="/content/runs/detect/predict/img1.jpg"))
```

```python
for r in results:
    print(f"Detected objects: {len(r.boxes)}")
    for box in r.boxes:
        print(
            f"Class: {int(box.cls[0])}, "
            f"Confidence: {float(box.conf[0]):.2f}"
        )
```

```python
results = model.predict(
    source="/content/img1.jpg",
    conf=0.25,
    save=True,
    show_labels=False
)
```

```python
from IPython.display import Image, display

results[0].save(filename="/content/result.jpg")
display(Image(filename="/content/result.jpg"))
```

```python
print("Total products detected:", len(results[0].boxes))
```

```python
for conf in [0.25, 0.40, 0.50, 0.60, 0.70]:
    r = model.predict(
        source="/content/img1.jpg",
        conf=conf,
        verbose=False
    )
    print(f"Confidence {conf}: {len(r[0].boxes)} detections")
```
