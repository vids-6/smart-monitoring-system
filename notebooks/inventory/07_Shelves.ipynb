```python
import json
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
file_name = list(uploaded.keys())[0]

with open(file_name, "r") as f:
    detections = json.load(f)

print("Detection data loaded!")
print("Total detections:", len(detections))
```

```python
uploaded_image = files.upload()
```

```python
image_name = list(uploaded_image.keys())[0]

image = cv2.imread(
    os.path.join("/content", image_name)
)

if image is None:
    raise ValueError("Could not load image.")

image_rgb = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)

print("Image loaded successfully!")
print("Width :", image.shape[1])
print("Height:", image.shape[0])

plt.figure(figsize=(14, 8))
plt.imshow(image_rgb)
plt.axis("off")
plt.title("Shelf Image")
plt.show()
```

```python
plt.figure(figsize=(14, 8))

x = [
    d["center_x"]
    for d in detections
]

y = [
    d["center_y"]
    for d in detections
]

plt.scatter(x, y, s=50)

plt.gca().invert_yaxis()

plt.xlabel("X coordinate")
plt.ylabel("Y coordinate")
plt.title("Detected Product Centers")

plt.grid(True)
plt.show()
```

```python
y_positions = sorted(
    d["center_y"]
    for d in detections
)

print("Sorted Y coordinates:\n")

for i, y in enumerate(y_positions):
    print(f"{i+1:2d}. {y:.2f}")
```

```python
y_gaps = np.diff(y_positions)

print("Y-coordinate gaps:\n")

for i, gap in enumerate(y_gaps):
    print(
        f"{i+1:2d} → {i+2:2d}: "
        f"{gap:.2f}"
    )
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    range(1, len(y_gaps) + 1),
    y_gaps,
    marker="o"
)

plt.xlabel("Consecutive detection pair")
plt.ylabel("Y-coordinate gap")
plt.title("Vertical Gaps Between Product Detections")

plt.grid(True)
plt.show()
```

```python
median_gap = np.median(y_gaps)

threshold = max(
    median_gap * 5,
    30
)

print("Median Y gap:", round(median_gap, 2))
print("Shelf boundary threshold:", round(threshold, 2))
```

```python
boundary_indices = [
    i
    for i, gap in enumerate(y_gaps)
    if gap > threshold
]

print("Potential shelf boundaries:")

for index in boundary_indices:
    print(
        f"Between detection "
        f"{index + 1} and {index + 2}"
    )
```

```python
sorted_detections = sorted(
    detections,
    key=lambda d: d["center_y"]
)

shelves = []
current_shelf = []

for i, detection in enumerate(sorted_detections):

    current_shelf.append(detection)

    # Check whether next detection starts a new shelf
    if i < len(sorted_detections) - 1:

        current_y = detection["center_y"]
        next_y = sorted_detections[i + 1]["center_y"]

        gap = next_y - current_y

        if gap > threshold:
            shelves.append(current_shelf)
            current_shelf = []

# Add final shelf
if current_shelf:
    shelves.append(current_shelf)

print("Automatically detected shelves:", len(shelves))
```

```python
for shelf in shelves:

    shelf.sort(
        key=lambda d: d["center_x"]
    )

print("Products sorted left-to-right within each shelf.")
```

```python
print("SHELF SUMMARY")
print("=" * 40)

for i, shelf in enumerate(shelves, start=1):

    average_y = np.mean([
        d["center_y"]
        for d in shelf
    ])

    print(
        f"Shelf {i}: "
        f"{len(shelf)} products | "
        f"average Y = {average_y:.1f}"
    )

print("\nTotal products:",
      sum(len(shelf) for shelf in shelves))
```

```python
shelf_image = image.copy()

for i, shelf in enumerate(shelves, start=1):

    y_values = [
        d["center_y"]
        for d in shelf
    ]

    average_y = int(
        np.mean(y_values)
    )

    # Draw shelf center line
    cv2.line(
        shelf_image,
        (0, average_y),
        (shelf_image.shape[1], average_y),
        (255, 0, 0),
        3
    )

    # Shelf label
    label = (
        f"Shelf {i} "
        f"({len(shelf)} products)"
    )

    cv2.putText(
        shelf_image,
        label,
        (20, max(30, average_y - 15)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 0, 0),
        2
    )

shelf_rgb = cv2.cvtColor(
    shelf_image,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(16, 10))
plt.imshow(shelf_rgb)
plt.axis("off")
plt.title("Automatically Detected Shelf Rows")
plt.show()
```

```python
shelf_data = []

for shelf_id, shelf in enumerate(
    shelves,
    start=1
):

    shelf_info = {
        "shelf_id": shelf_id,
        "product_count": len(shelf),
        "average_y": float(
            np.mean([
                d["center_y"]
                for d in shelf
            ])
        ),
        "products": shelf
    }

    shelf_data.append(shelf_info)

output_path = "/content/shelves.json"

with open(output_path, "w") as f:
    json.dump(
        shelf_data,
        f,
        indent=4
    )

print("Shelf data saved to:")
print(output_path)
```
